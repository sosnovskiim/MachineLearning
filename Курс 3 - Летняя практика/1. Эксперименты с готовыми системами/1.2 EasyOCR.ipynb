```python
import matplotlib.patches as patches
import matplotlib.pyplot as plt
import warnings
from PIL import Image
from time import time

import easyocr

reader = easyocr.Reader(["ru", "en"], gpu=False)
```

```python
def show_image(image_path):
    image = Image.open(image_path)
    plt.imshow(image)
    plt.axis("off")
    plt.show()

def recognize_text(image_path):
    start_time = time()
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        result = reader.readtext(image_path)
    end_time = time()

    elapsed_time = f"{end_time - start_time:.3f}"
    text = "\n".join([item[1] for item in result])

    return text, elapsed_time, result

def show_bounding_boxes(image_path, result):
    image = Image.open(image_path)
    
    ax = plt.gca()
    ax.axis("off")
    ax.imshow(image)
    
    for (bbox, _, _) in result:
        rect = patches.Polygon(bbox, linewidth=1.5, edgecolor="red", facecolor="none")
        ax.add_patch(rect)
    
    plt.show()

def test_on_image(image_path):
    print(f"Изображение: {image_path}")
    show_image(image_path)

    text, elapsed_time, result = recognize_text(image_path)

    print("Изображение с обнаруженным текстом:")
    show_bounding_boxes(image_path, result)

    sep = "-" * 50
    print(
        f"Распознанный текст (время: {elapsed_time} секунд):",
        sep, text, sep, sep="\n"
    )
```
